```python
import requests
import json

API_TOKEN = "PASTE_YOUR_HUGGINGFACE_API_KEY"

headers = {
    "Authorization": f"Bearer {API_TOKEN}"
    }

API_URL = "https://router.huggingface.co/hf-inference/models/mrm8488/bert-tiny-finetuned-fake-news-detection"

def check_fake_news(text):
        payload = {
                "inputs": text
                    }
        response = requests.post(API_URL, headers=headers, json=payload)
        return response.json()


news = input("Enter a news article or claim:\n")


result = check_fake_news(news)

predictions = result[0]
best = max(predictions, key=lambda x: x['score'])

label_map = {
    "LABEL_1": "FAKE NEWS",
        "LABEL_0": "REAL NEWS"
        }

final_label = label_map.get(best['label'], best['label'])
confidence = round(best['score'] * 100, 2)

print("\n🔍 Prediction Result")
print("----------------------")
print("Statement :", news)
print("Prediction:", final_label)
print("Confidence:", confidence, "%")
print("\nRaw AI Output:\n", result)
```
